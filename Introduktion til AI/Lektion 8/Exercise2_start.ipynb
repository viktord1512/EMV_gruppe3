# Prepare data

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Read data
data = pd.read_csv('titanic.csv')
data.head()
```

```python
# Preprocessing
# Drop columns we don't need
data.drop(['PassengerId','Name','Cabin','Ticket','Embarked','Fare'],axis=1,inplace=True) 

# Drop nan rows for the dataset
data.dropna(inplace=True) 

# Replace strings with 0 and 1
data['Sex']=data['Sex'].map({'male': 1, 'female': 0})

data.head()
```

```python
# Define data
X = data[['Pclass','Sex','Age','SibSp','Parch']].values
y = data[['Survived']].values

# Split data into training and test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Initialize the model
model = DecisionTreeClassifier(
    min_samples_leaf=4
)

# Create a Decision tree classifier
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Calculate accuracy of the model with %
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Initialize the model
model = RandomForestClassifier(
    min_samples_leaf=5,
    min_samples_split=20,
    n_estimators=300
)

# Create a Random forest classifier
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Calculate accuracy of the model with %
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
```
